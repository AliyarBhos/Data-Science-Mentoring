# **Matplotlib**

```python
import matplotlib.pyplot as plt
```

## **Histogram**

```python
df_pitt = df[df['city']=="Pittsburgh"]

df_vegas = df[df['city']=="Las Vegas"]

df_vegas
```

```python
df_pitt = df[df["city"] == "Pittsburgh"]
df_vegas = df[df["city"] == "Las Vegas"]
series_pitt = df_pitt["stars"]
series_vegas = df_vegas["stars"]
```

```python
plt.hist(
    series_pitt,  #star ratings for Pittsburg
    alpha = 0.3,  #transparency

    color = 'red',  #histogram color

    label = "Pittsburgh", #label
    bins='auto',
)
plt.legend()
plt.show()
```

```python
plt.hist(series_vegas, alpha = 0.3, color = 'red', label = 'Vegas', bins = 'auto')
plt.hist(series_pitt, alpha = 0.3, color = 'blue', label = "Pittsburgh", bins='auto')
plt.legend()
plt.show()
```

```python
plt.hist(series_pitt, alpha = 0.3, color = 'blue', label = "Pittsburgh", bins='auto')
plt.hist(series_vegas, alpha = 0.3, color = 'red', label = 'Vegas', bins = 'auto')
plt.xlabel("Rating")
plt.ylabel("Number of Rating Score")
plt.legend(loc='best')
plt.title("Review distribution of Pittsburgh and Las Vegas")
plt.show()
```

```python
plt.hist([series_pitt,series_vegas],
         alpha = 0.7,
         color = ['blue', 'red'],
         label = ["Pittsburgh", 'Las Vegas'],
         bins='auto')
plt.xlabel("Rating")
plt.ylabel("Number of Rating Score")
plt.legend(loc='best')
plt.title("Review distribution of Pittsburgh and Las Vegas")
plt.show()
```

## **Scatterplot**

```python
df_health = df[df['category_0'] == "Health & Medical"]

df_fast = df[df["category_0"] == "Fast Food"]

df_brunch = df[df['category_0'] == "Breakfast & Brunch"]
```

```python
plt.scatter(df_health['stars'], df_health['review_count'],
            marker = 'o',
            color = 'r',
            alpha = 0.7,
            s = 124,
            label = ["Health & Medical"])

plt.scatter(df_fast['stars'], df_fast['review_count'],
            marker = 'h',
            color = 'b',
            alpha = 0.7,
            s = 124,
            label = ["Fast Food"])

plt.scatter(df_brunch['stars'], df_brunch['review_count'],
            marker = '^',
            color = 'g',
            alpha = 0.7,
            s = 124,
            label = ["Fast Food"])

plt.xlabel("Rating")
plt.ylabel('Review Count')

plt.legend(loc = "upper left")

plt.show()
```

<h2 style="color:#7c3aed; margin-top:0;">🟪 QUESTION 3 · LOOPS & CONDITIONS</h2>
<h3>🔢 Find Even and Odd Numbers</h3>
<b>Task:</b> Ask the user for an integer <code>n</code>. Print all even numbers from 1 to <code>n</code>, followed by all odd numbers.
<h4 style="color:#0f766e;">🇦🇿 Cüt və tək ədədləri tapın</h4>
<b>Tapşırıq:</b> İstifadəçidən tam ədəd <code>n</code> daxil etməsini istəyin. 1-dən <code>n</code>-ə qədər olan cüt ədədləri, daha sonra isə tək ədədləri ekrana çıxarın.
<div style="background:#f5f3ff; padding:10px; border-radius:7px;">
<b style="color:#6d28d9;">Example / Nümunə</b>
<pre>Enter n: 10

Even numbers: 2 4 6 8 10
Odd numbers: 1 3 5 7 9</pre>
</div>
</div>


```python
n = int(input("Enter n: "))

print("Even numbers:", end=" ")

for i in range(1, n + 1):
    if i % 2 == 0:
        print(i, end=" ")

print()

print("Odd numbers:", end=" ")

for i in range(1, n + 1):
    if i % 2 != 0:
        print(i, end=" ")
```

<h2 style="color:#be185d; margin-top:0;">🟥 QUESTION 4 · NESTED LOOPS & LISTS</h2>
<h3>🧮 Sum of Each Matrix Row</h3>
<b>Task:</b> Given a two-dimensional list representing a matrix, calculate the sum of each row and store the results in a new list.
<h4 style="color:#0f766e;">🇦🇿 Matrisin hər bir sətrinin cəmini hesablayın</h4>
<b>Tapşırıq:</b> Verilmiş ikiölçülü siyahıda matrisin hər bir sətrinin elementlərinin cəmini hesablayın və nəticələri yeni siyahıda saxlayın.
<div style="background:#fdf2f8; padding:10px; border-radius:7px;">
<b style="color:#9d174d;">Input / Giriş</b>
<pre>matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]</pre>
<b style="color:#9d174d;">Expected output / Gözlənilən nəticə</b>
<pre>[6, 15, 24]</pre>
</div>
</div>


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

row_sums = []

for row in matrix:
    row_sums.append(sum(row))

print(row_sums)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

row_sums = []

for row in matrix:
    total = 0

    for number in row:
        total += number

    row_sums.append(total)

print(row_sums)
```

<h2 style="color:#0e7490; margin-top:0;">🟦 QUESTION 5 · FUNCTIONS & *ARGS</h2>
<h3>📊 Create a Flexible Statistics Function</h3>
<b>Task:</b> Write <code>calculate_stats(*args)</code>, which accepts any number of numeric arguments and returns a dictionary containing the minimum, maximum, sum and average. Handle the case when no arguments are provided.
<h4 style="color:#0f766e;">🇦🇿 Çevik statistika funksiyası yaradın</h4>
<b>Tapşırıq:</b> İstənilən sayda ədədi arqument qəbul edən <code>calculate_stats(*args)</code> funksiyası yazın. Funksiya minimum, maksimum, cəm və orta qiyməti lüğət şəklində qaytarmalıdır. Heç bir arqument ötürülmədiyi halı da nəzərə alın.
<div style="background:#ecfeff; padding:10px; border-radius:7px;">
<b style="color:#155e75;">Expected output / Gözlənilən nəticə</b>
<pre>{
    "min": 10,
    "max": 50,
    "sum": 150,
    "average": 30.0
}</pre>
</div>
</div>


```python
def calculate_stats(*args):
    if not args:
        return None

    result = {
        "min": min(args),
        "max": max(args),
        "sum": sum(args),
        "average": sum(args) / len(args)
    }

    return result


print(calculate_stats(10, 20, 30, 40, 50))
```

<h2 style="color:#c2410c; margin-top:0;">🟧 QUESTION 6 · EXCEPTIONS & FUNCTIONS</h2>
<h3>🧮 Build a Safe Calculator</h3>
<b>Task:</b> Create a function <code>calculate(a, b, operator)</code> that supports addition, subtraction, multiplication and division. Handle division by zero and unsupported operators using exceptions.
<h4 style="color:#0f766e;">🇦🇿 Təhlükəsiz kalkulyator yaradın</h4>
<b>Tapşırıq:</b> Toplama, çıxma, vurma və bölmə əməliyyatlarını dəstəkləyən <code>calculate(a, b, operator)</code> funksiyası yaradın. Sıfıra bölmə və dəstəklənməyən operator hallarında exception-lardan istifadə edərək xətaları idarə edin.
<div style="background:#fff7ed; padding:10px; border-radius:7px;">
<b style="color:#9a3412;">Example / Nümunə</b>
<pre>10 + 5 = 15
10 / 2 = 5.0
Error: Division by zero
Error: Unsupported operator</pre>
</div>
</div>


```python
def calculate(a, b, operator):

    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            raise ZeroDivisionError("Division by zero")
        return a / b
    else:
        raise ValueError("Unsupported operator")


tests = [
    (10, 5, "+"),
    (10, 2, "/"),
    (10, 0, "/"),
    (10, 5, "%")
]

for a, b, operator in tests:
    try:
        result = calculate(a, b, operator)
        print(f"{a} {operator} {b} = {result}")

    except (ZeroDivisionError, ValueError) as e:
        print("Error:", e)
```

<h2 style="color:#4f46e5; margin-top:0;">🟪 QUESTION 7 · LIST COMPREHENSION</h2>
<h3>🔄 Filter and Transform Numbers</h3>
<b>Task:</b> Given a list of integers, create a new list containing the squares of only the even numbers using a list comprehension.
<h4 style="color:#0f766e;">🇦🇿 Ədədləri seçin və çevrilməsini həyata keçirin</h4>
<b>Tapşırıq:</b> Tam ədədlərdən ibarət verilmiş siyahıdan yalnız cüt ədədlərin kvadratlarını list comprehension vasitəsilə yeni siyahıda yaradın.
<div style="background:#eef2ff; padding:10px; border-radius:7px;">
<b style="color:#4338ca;">Input / Giriş</b>
<pre>numbers = [1, 2, 3, 4, 5, 6, 7, 8]</pre>
<b style="color:#4338ca;">Expected output / Gözlənilən nəticə</b>
<pre>[4, 16, 36, 64]</pre>
</div>
</div>


```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

result = [number ** 2 for number in numbers if number % 2 == 0]

print(result)
```

<h2 style="color:#15803d; margin-top:0;">🟩 QUESTION 8 · OBJECT-ORIENTED PROGRAMMING</h2>
<h3>👤 Create a Simple Student Class</h3>
<b>Task:</b> Create a class named <code>Student</code> with two attributes: <code>name</code> and <code>age</code>. Create one student object and display its attributes.
<h4 style="color:#0f766e;">🇦🇿 Sadə Student sinfi yaradın</h4>
<b>Tapşırıq:</b> <code>Student</code> adlı sinif yaradın. Sinifdə <code>name</code> və <code>age</code> adlı iki atribut olsun. Bir tələbə obyekti yaradın və onun atributlarını ekrana çıxarın.
<div style="background:#f0fdf4; padding:10px; border-radius:7px;">
<b style="color:#166534;">Example / Nümunə</b>
<pre>Student name: Ali
Student age: 20</pre>
</div>
</div>

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Ali", 20)

print("Student name:", student1.name)
print("Student age:", student1.age)
```

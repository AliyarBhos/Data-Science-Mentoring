```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("pokemon.csv")

print("First 10 rows:")
print(df.head(10))

print("\nLast 5 rows:")
print(df.tail())

print("\nDataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

df.info()

print(df.describe())

print("\nUnique Pokemon names:", df["Name"].nunique())

print("\nUnique primary types:", df["Type 1"].nunique())
print(df["Type 1"].unique())

print(sorted(df["Generation"].dropna().unique()))
```

```python

print("\nMissing values:")
print(df.isnull().sum())

print("\nMissing value percentages:")
print((df.isnull().mean() * 100).round(2))

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)

print("\nMissing Type 2 values:", df["Type 2"].isnull().sum())

stats = [
    "HP", "Attack", "Defense",
    "Sp. Atk", "Sp. Def", "Speed"
]

print("\nNegative stat values:")
print((df[stats] < 0).sum())

print("\nMissing names:", df["Name"].isnull().sum())

print("\nGeneration values:")
print(df["Generation"].value_counts(dropna=False).sort_index())
```

```python
high_attack = df.loc[df["Attack"] > 100]
print("\nPokemon with Attack > 100:")
print(high_attack[["Name", "Type 1", "Attack"]])

fast_defensive = df.loc[
    (df["Speed"] > 100) & (df["Defense"] > 80)
]
print("\nSpeed > 100 and Defense > 80:")
print(fast_defensive[["Name", "Speed", "Defense"]])

water_pokemon = df.loc[df["Type 1"] == "Water"]
print("\nWater Pokemon:", len(water_pokemon))

legendary = df.loc[df["Legendary"] == True]
print("\nLegendary Pokemon:")
print(legendary[["Name", "Total", "Legendary"]])

generation_1 = df.loc[df["Generation"] == 1]
print("\nGeneration 1 Pokemon:", len(generation_1))

top_total = df.nlargest(10, "Total")
print("\nTop 10 by Total:")
print(top_total[["Name", "Total"]])

slowest = df.nsmallest(10, "Speed")
print("\nBottom 10 by Speed:")
print(slowest[["Name", "Speed"]])

fire_attack = df.loc[
    (df["Type 1"] == "Fire") & (df["Attack"] > 90)
]
print("\nFire Pokemon with Attack > 90:")
print(fire_attack[["Name", "Attack"]])

no_secondary_type = df.loc[df["Type 2"].isna()]
print("\nPokemon without secondary type:", len(no_secondary_type))

fast_legendary = df.loc[
    (df["Legendary"] == True) & (df["Speed"] > 90)
]
print("\nFast legendary Pokemon:")
print(fast_legendary[["Name", "Speed", "Total"]])

print("\nNumber of Fire Pokemon with Attack > 90:", len(fire_attack))
print("Number of fast legendary Pokemon:", len(fast_legendary))
```

```python
df["Average_Stat"] = df[stats].mean(axis=1)

df["Offensive_Power"] = df["Attack"] + df["Sp. Atk"]

df["Defensive_Power"] = df["Defense"] + df["Sp. Def"]

df["Strength_Category"] = pd.cut(
    df["Total"],
    bins=[-np.inf, 300, 500, np.inf],
    labels=["Weak", "Moderate", "Strong"],
    right=False
)

df["Has_Secondary_Type"] = df["Type 2"].notna()

print("\nStrength category counts:")
print(df["Strength_Category"].value_counts().sort_index())

print("\nHighest Offensive Power:")
print(
    df.nlargest(1, "Offensive_Power")[
        ["Name", "Attack", "Sp. Atk", "Offensive_Power"]
    ]
)

print("\nHighest Defensive Power:")
print(
    df.nlargest(1, "Defensive_Power")[
        ["Name", "Defense", "Sp. Def", "Defensive_Power"]
    ]
)

print("\nAverage stat by legendary status:")
print(df.groupby("Legendary", observed=True)["Average_Stat"].mean())
```

```python
type_counts = df.groupby("Type 1").size().sort_values(ascending=False)
print("\nPokemon count by type:")
print(type_counts)

avg_attack = df.groupby("Type 1")["Attack"].mean().sort_values(
    ascending=False
)
print("\nAverage Attack by type:")
print(avg_attack)

avg_defense = df.groupby("Type 1")["Defense"].mean().sort_values(
    ascending=False
)
print("\nAverage Defense by type:")
print(avg_defense)

avg_speed_gen = df.groupby("Generation")["Speed"].mean()
print("\nAverage Speed by generation:")
print(avg_speed_gen)

print("\nLegendary counts:")
print(df["Legendary"].value_counts(dropna=False))

print("\nAverage Total by legendary status:")
print(df.groupby("Legendary")["Total"].mean())

avg_total_type = (
    df.groupby("Type 1")["Total"]
    .mean()
    .sort_values(ascending=False)
)
print("\nTop 5 types by average Total:")
print(avg_total_type.head(5))

hp_summary = df.groupby("Generation")["HP"].agg(
    ["min", "max", "mean"]
)
print("\nHP statistics by generation:")
print(hp_summary)

print("\nSecondary type presence:")
print(df["Has_Secondary_Type"].value_counts())

attack_by_type_legendary = df.groupby(
    ["Type 1", "Legendary"]
)["Attack"].mean().unstack()

print("\nAverage Attack by type and legendary status:")
print(attack_by_type_legendary)
```

```python
pivot_total = pd.pivot_table(
    df,
    values="Total",
    index="Type 1",
    columns="Legendary",
    aggfunc="mean",
    observed=False
)
print("\nAverage Total by type and legendary status:")
print(pivot_total)

pivot_attack = pd.pivot_table(
    df,
    values="Attack",
    index="Generation",
    columns="Type 1",
    aggfunc="mean",
    observed=False
)
print("\nAverage Attack by generation and type:")
print(pivot_attack)

pivot_count = pd.pivot_table(
    df,
    index="Generation",
    columns="Legendary",
    values="Name",
    aggfunc="count",
    fill_value=0,
    observed=False
)
print("\nPokemon count by generation and legendary status:")
print(pivot_count)

pivot_stats = pd.pivot_table(
    df,
    values=["HP", "Attack", "Defense"],
    index="Generation",
    aggfunc="mean",
    observed=False
)
print("\nAverage stats by generation:")
print(pivot_stats)

legendary_by_type = df.groupby("Type 1")["Legendary"].sum()
no_legendary_types = legendary_by_type[legendary_by_type == 0]

print("\nTypes with no legendary Pokemon:")
print(no_legendary_types.index.tolist())
```

```python
plt.figure(figsize=(12, 6))
type_counts.sort_values().plot(kind="barh")
plt.title("Number of Pokemon by Primary Type")
plt.xlabel("Number of Pokemon")
plt.ylabel("Primary Type")
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="Total", bins=25, kde=True)
plt.title("Distribution of Total Battle Stats")
plt.xlabel("Total")
plt.ylabel("Number of Pokemon")
plt.tight_layout()
plt.show()


plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x="Legendary", y="Total")
plt.title("Total Stats: Legendary vs Non-Legendary")
plt.xlabel("Legendary")
plt.ylabel("Total")
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Attack",
    y="Defense",
    hue="Legendary",
    alpha=0.8
)
plt.title("Attack vs Defense")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 8))
correlation = df[stats + ["Total"]].corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Pokemon Battle Stats")
plt.tight_layout()
plt.show()
```

```python
long_tracks = long_tracks.copy()

long_tracks["duration_min"] = long_tracks["duration_ms"] / 60000

long_tracks[
    ["track_name", "duration_ms", "duration_min"]
].head(10)
```

```python
cleaned = df.copy()
```

```python
cleaned["artists"] = cleaned["artists"].fillna("Unknown Artist")
```

```python
numeric_columns = cleaned.select_dtypes(include="number").columns

for column in numeric_columns:
    cleaned[column] = cleaned[column].fillna(
        cleaned[column].median()
    )
```

```python
cleaned = cleaned.drop_duplicates()
```

```python
print("Missing values after cleaning:")
print(cleaned.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(cleaned.duplicated().sum())
```

```python
print(cleaned["track_genre"].unique())

print("Number of genres:", cleaned["track_genre"].nunique())
```

```python
genre_counts = cleaned["track_genre"].value_counts()

print(genre_counts.head(10))
```

```python
genre_counts.head(10).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Top 10 Most Common Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
genre_popularity = (
    cleaned.groupby("track_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
)

print(genre_popularity.head(10))
```

```python
top_5_genres = genre_popularity.head(5)

print(top_5_genres)
```

```python
cleaned["duration_min"] = cleaned["duration_ms"] / 60000

cleaned[
    ["track_name", "duration_ms", "duration_min"]
].head(10)
```

```python
cleaned["popularity_level"] = np.where(
    cleaned["popularity"] >= 70,
    "Popular",
    "Less Popular"
)

cleaned[
    ["track_name", "popularity", "popularity_level"]
].head(10)
```

```python
popularity_pivot = pd.pivot_table(
    cleaned,
    values="popularity",
    index="track_genre",
    columns="popularity_level",
    aggfunc="mean"
)

popularity_pivot.head(10)
```

```python
genre_mapping = pd.DataFrame({
    "track_genre": ["rock", "pop", "jazz"],
    "genre_category": [
        "Rock Music",
        "Pop Music",
        "Jazz Music"
    ]
})

genre_mapping
```

```python
merged = cleaned.merge(
    genre_mapping,
    on="track_genre",
    how="left"
)

merged[
    ["track_name", "track_genre", "genre_category"]
].head(10)
```

```python
top_5_genres.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Top 5 Genres by Average Popularity")
plt.xlabel("Genre")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    cleaned["popularity"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    cleaned["danceability"],
    cleaned["popularity"],
    alpha=0.3
)

plt.title("Danceability vs Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

```python
top_common_genres = cleaned["track_genre"].value_counts().head(5).index

boxplot_data = [
    cleaned.loc[
        cleaned["track_genre"] == genre,
        "popularity"
    ].dropna()
    for genre in top_common_genres
]

plt.figure(figsize=(10, 5))

plt.boxplot(
    boxplot_data,
    tick_labels=top_common_genres
)

plt.title("Popularity Distribution Across Five Genres")
plt.xlabel("Genre")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

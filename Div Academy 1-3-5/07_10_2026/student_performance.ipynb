### messy

```python
messy_students = students.copy()

messy_students.loc[0, "math score"] = np.nan
messy_students.loc[1, "gender"] = np.nan
messy_students.loc[2, "lunch"] = np.nan

messy_students = pd.concat(
    [messy_students, messy_students.iloc[[3]]],
    ignore_index=True
)

messy_students.head()
```

```python
messy_students = messy_students.drop_duplicates() 

messy_students
```

```python
messy_students["math score"] = messy_students[ 
    "math score" 
].fillna( messy_students["math score"].mean() )
```

```python
messy_students["gender"] = messy_students[
    "gender"
].fillna("unknown")
messy_students["lunch"] = messy_students[
    "lunch"
].fillna("unknown")
messy_students.isnull().sum()
```

```python
messy_students = messy_students.dropna(
    subset=["gender"]
)

messy_students
```

```python
students["math score"].plot.hist(bins=10) 

plt.xlabel("Math Score") 
plt.ylabel("Number of Students") 
plt.title("Distribution of Math Scores") 

plt.show()
```

```python
students[ students["math score"] > 80 ].plot.hist( y="math score", bins=10) 

plt.xlabel("Math Score") 
plt.ylabel("Number of Students") 
plt.title("Math Scores Above 80") 
plt.show()
```

```python
female_scores = students[
    students["gender"] == "female"
]["math score"]

male_scores = students[
    students["gender"] == "male"
]["math score"]
plt.hist(
    female_scores,
    bins="auto",
    alpha=0.6,
    label="Female"
)

plt.hist(
    male_scores,
    bins="auto",
    alpha=0.6,
    label="Male"
)

plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.title("Math Scores by Gender")
plt.legend()
plt.show()
```

```python
plt.scatter( students["math score"], students["reading score"], alpha=0.5 ) 
plt.xlabel("Math Score") 
plt.ylabel("Reading Score") 
plt.title("Math Score vs Reading Score") 

plt.show()
```

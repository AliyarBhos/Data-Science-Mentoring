```python
text = input("Enter text: ")

count = 0

for char in text:
    if char != " ":
        count += 1

print("Number of characters:", count)
```

```python
numbers = [12, 45, 7, 89, 23, 56]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

```python
grades = {
    "Ali": 85,
    "Leyla": 62,
    "Murad": 91,
    "Aysel": 68
}

for name, grade in grades.items():
    if grade >= 70:
        print(name)
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


print(is_prime(7))
print(is_prime(10))
```

```python
numbers = [4, 12, 7, 18, 21, 24, 9, 30]

result = [number for number in numbers if number > 10 and number % 2 == 0]

print(result)
```

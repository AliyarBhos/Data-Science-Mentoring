```python
messy_data = {
    "name": ["Ali", "Leyla", "Murad", "Ali", "Nigar"],
    "age": [21, 24, None, 21, 26],
    "city": ["Baku", "Ganja", "Baku", "Baku", None]
}

students = pd.DataFrame(messy_data)
```

```python
print(students.isnull().sum())
print(students.duplicated())
```

```python
students = students.drop_duplicates()

students["age"] = students["age"].fillna(
    students["age"].mean()
)
```

```python
students = students.dropna(subset=["city"])

print(students)
```
